```python
data["test_time"].describe()
```

```python
sample_ids = data["subject"].sample(3, random_state=42)

for pid in sample_ids:
    tmp = data[data["subject"] == pid]
    plt.plot(tmp["test_time"], tmp["motor_updrs"], label=f"Patient {pid}")

plt.legend()
plt.xlabel("Time")
plt.ylabel("Motor_updrs")
plt.show()
```

It can be seen that Patient 39 and patient 2 have an increase in motor_updrs with time. Patient 32 shows improvements and decrease in motor_updrs over time

CORRELATION ANALYSIS

```python
corr_analysis = data.corr(numeric_only=True)
```

```python
heat_map = corr_analysis["motor_updrs"].sort_values(ascending=True).head(10)
```

```python
top_features = corr_analysis["motor_updrs"].abs().sort_values(ascending=False).head(12).index

sns.heatmap(data[top_features].corr(), cmap="coolwarm", center=0)
plt.show()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

X = data[features]
vif = pd.DataFrame()
vif["feature"] = X.columns
vif["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
vif.sort_values("VIF", ascending=False)
```

High multicollinearity was observed among jitter and shimmer-related features (VIF > 80).
